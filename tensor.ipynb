```python
# inplace operations
m = torch.rand(2,3)
n = torch.rand(2,3)

print(m)
print(n)
```

```python
# for each operation performed on tensors, a new tensor is formed. what i want is if i add m with n, value should be stored in m itself.
m.add_(n)
print(m)
print(n)
```

```python
# similarly you can do the same with relu_...this underscore is inplace operation
```
