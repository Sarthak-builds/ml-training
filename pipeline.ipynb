```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace = True)
```

```python
#  Train Test SPlit
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size = 0.2, random_state=42)
# random state makes the same train test split reproducible each time you run the code
# iloc means integer location based indexing that lets you select rows and columns using there indexes.
```

```python
# scaling puts the features onto a comparable scale
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
# fit tranform means learn the scaling params from the data and scale the data
X_test = scaler.transform(X_test)
```

```python
#  label encoding
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
X_test_tensor = torch.from_numpy(X_test)
y_test_tensor = torch.from_numpy(y_test)
y_train_tensor = torch.from_numpy(y_train)
X_train_tensor = torch.from_numpy(X_train)
```

```python
X_train_tensor.shape
```

```python
class MySimpleNN():
    def __init__(self, X):

        self.weights = torch.rand(X.shape[1],1, dtype= torch.float64, requires_grad = True)
        self.bias = torch.zeros(1, dtype = torch.float64, requires_grad = True)


    def forward(self, X):
        z = torch.matmul(X, self.weights) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred

    def loss_function(self, y_pred, y_train_tensor):
        # clamp 
        epsilon = 1e-7
        y_pred = torch.clamp(y_pred, epsilon, 1-epsilon)
        loss = -(y_train_tensor * torch.log(y_pred) + (1 -  y_train_tensor) * torch.log(1 - y_pred)).mean()
        return loss
```

```python
# important parameters
learning_rate = 0.1
epochs = 100000
```

```python
# create model
model = MySimpleNN(X_train_tensor)

# define the loop
for epoch in range(epochs):

    # forward pass
    y_pred = model.forward(X_train_tensor)

    # loss calc
    loss = model.loss_function(y_pred, y_train_tensor)

    # backward pass
    loss.backward()

    # params update
    with torch.no_grad():
      model.weights -= learning_rate * model.weights.grad
      model.bias -=  learning_rate * model.bias.grad

    # zero gradients
    model.weights.grad.zero_()
    model.bias.grad.zero_()
    
    print(f'Epoch: { epoch + 1}, Loss: { loss.item()}')



  
```

```python
# model evaluation
with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred = (y_pred >  0.5).float()
    accuracy = ( y_pred == y_test_tensor).float().mean()
print(f'Accuracy: {accuracy.item()}')    

```
